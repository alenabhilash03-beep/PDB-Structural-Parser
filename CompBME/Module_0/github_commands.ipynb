### **Step 4. Add and commit your changes**
- You have now just added a new folder into your local repository folder! Every time you make any changes to your files, or add new files, you should add and commit your changes.
- **git status**: lists any untracked changes you just made to your folder.
- **git add**: adds your changes to the "stage"
- **git commit -m "message"**: fully commits those changes with a message
- **git commit -a**: if you want to add and commit in one line.
- You always need to write a message with your commit.

```python
# change current directory to the local repo folder
%cd /Users/lln5zuj/Library/CloudStorage/OneDrive-UniversityofVirginia/Graduate School/BME 2315 Fall 2026/Module 0/BME2315_repo 

!git status # displays any untracked changes to be added
```

```python
!git commit -m "added python_practice.py" # write a descriptive commit message
```

### **Step 6. Push your changes to Git or GitHub (remote)**
- To finalize your changes, push them internally to Git (now it has fully tracked your file) using **git push**
- To push them to your **main** branch in your GitHub repository, use **git push origin main**
- **Make sure you are signed into GitHub on VSCode**

```python
!git push # pushes your changes to Git, it is now tracked
!git push origin main # pushes your changes to the main branch on your GitHub repository
```

### **You are done**! This is the process to make changes to your local files, add them, commit them (with a message), and push them to Git (interally) and/or GitHub (for others to see).

**Overall workflow:**
1. Clone your GitHub repository to a local directory on your computer (you may be prompted by VSCode to sign into GitHub if you haven't done so before) **(git clone)**
2. Add your files to the local repo folder and begin making changes.
3. Add your files to the "staging area" **(git add)**
4. Commit your files with a written message **(git commit -m "message")**
5. Push your files to Git **(git push)**
6. Push your files to GitHub (main branch) **(git push origin main)**
